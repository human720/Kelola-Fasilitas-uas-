```python
# =========================
# MODUL FASILITAS
# =========================
def lihat_fasilitas():
    print("\n=== DAFTAR FASILITAS ===")
    with koneksi() as conn:
        cur = conn.cursor()
        cur.execute("""
            SELECT id, nama_fasilitas, jenis, harga_per_jam
            FROM fasilitas ORDER BY id
        """)
        data = cur.fetchall()

    if not data:
        print("Belum ada fasilitas.")
        return

    print("-" * 75)
    print(f"{'ID':<5}{'Nama Fasilitas':<30}{'Jenis':<15}{'Harga/Jam':>15}")
    print("-" * 75)
    for item in data:
        print(f"{item[0]:<5}{item[1]:<30}{item[2]:<15}Rp{item[3]:>13,}")
    print("-" * 75)


def tambah_fasilitas():
    print("\n=== TAMBAH FASILITAS ===")
    nama = input("Nama fasilitas : ").strip()
    jenis = input("Jenis          : ").strip()

    try:
        harga = int(input("Harga per jam  : Rp"))
        if not nama or not jenis or harga <= 0:
            raise ValueError
    except ValueError:
        print("Data tidak valid. Harga harus berupa angka lebih dari 0.")
        return

    with koneksi() as conn:
        conn.execute("""
            INSERT INTO fasilitas (nama_fasilitas, jenis, harga_per_jam)
            VALUES (?, ?, ?)
        """, (nama, jenis, harga))
        conn.commit()
    print("Fasilitas berhasil ditambahkan.")
```
